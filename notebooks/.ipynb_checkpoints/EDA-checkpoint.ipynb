```python
import pandas as pd
import numpy as np
```

## Articles

```python
articles = pd.read_json('../data/articles.json')
```

```python
for i in range(articles.shape[0]):
    print('title: ')
    print(articles.iloc[i]['title'])
    print('-' * 50)
    print('article: ')
    print(articles.iloc[i]['text'])
    print('#' * 50)
```

## Questions with right answers

```python
questions = pd.read_json('../data/questions.json')
truth = pd.read_json('../data/ground_truth.json')
```

```python
questions['answer'] = truth['answer']
```

```python
for i in range(10):
    print('question: ')
    print(questions.iloc[i]['question'])
    print('-' * 50)
    print('answer: ')
    print(questions.iloc[i]['answer'])
    print('#' * 50)    
```
